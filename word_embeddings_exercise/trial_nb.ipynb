```python
from transformers import DistilBertTokenizer, DistilBertModel



tokenizer = DistilBertTokenizer.from_pretrained('distilbert-base-uncased')

with open("Word_Embeddings_hands_on\corpus.txt", "r", encoding="utf-8") as f:
    text = f.read()

tokens = tokenizer(text, padding=True, truncation=True, return_tensors='pt')

```

```python
type(tokens)
```

```python
model = DistilBertModel.from_pretrained('distilbert-base-uncased')

model.eval()
```

```python
from torch.utils.data import DataLoader, TensorDataset
import torch

embeddings_list = []
dataset = TensorDataset(tokens['input_ids'], tokens['attention_mask'])
with torch.no_grad():
    for batch in DataLoader(dataset, batch_size=10):
        outputs = model(tokens['input_ids'], tokens['attention_mask'])
        embeddings = outputs.last_hidden_state
        sentence_emb = embeddings.mean(dim=1)
        embeddings_list.append(sentence_emb.cpu().numpy())
```

```python

import numpy as np
import pandas as pd

embeddings_np = np.vstack(embeddings_list)

embeddings_np.size

embeddings_df = pd.DataFrame(embeddings_np)

embeddings_df.head()

# from sklearn.decomposition import PCA

# pca = PCA(n_components=1)

# reduced_emb = pca.fit_transform(embeddings_np)
```
